# Getting started 

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from pyvasplot import PyVASP
from pyvasplot.plotting import plot_bands, plot_kpath, plot_structure
```

```python
path = Path("data/Sb_111_GGA_0014")

dft = PyVASP(
    path,
    calculation_type="BS_KPATH",
    sub_path="BS_MGKM",
)

dft.load()
print(dft)
```

## Show the unit cell with atoms

```python
plot_structure(dft)
plt.show()
```

## Plot the calculation path

```python
plot_kpath(dft)

plt.show()
```

## Plot the band structure

```python
plot_bands(dft)
plt.show()
```

## Plot a section of calculated band structure with symmetry M-G-M

```python
fig, ax = plt.subplots(1, 1, figsize=(3, 4))

sec_MGM = {'k_section':0, 'k_mirror':True, 'k_flip':True}

plot_bands(dft, ax=ax, **sec_MGM)
ax.set_title('M - G - M')
ax.set_xlabel('k (1/A)')
plt.show()
```
